```python
from time import perf_counter
import sys
sys.set_int_max_str_digits(0)
t0 = perf_counter()
arr = [1]
print(len(arr))
def factorials(x, whole_array):
    if x == 1: arr[x] = 1
    while(len(arr) < x):
        arr.append(arr[len(arr) - 1] * (len(arr) + 1))
    return arr[0:x] if whole_array else arr[x-1]

print(factorials(100000, False))
t1 = perf_counter()
print(f"Time taken = {t1 - t0:.6f} s")
```

```python
t2 = perf_counter()
print(factorials(99997, False))
t3 = perf_counter()
print(f"Time taken = {t3 - t2:.6f} s")
```

```python
def doubling_time(max_growth_factor=100):
    def day(growth_factor):
        return np.log(growth_factor)/np.log(1.01)
    n_max = int((np.log(max_growth_factor)/np.log(2))//1 + 1)
    print(f"n_max for max_growth_factor {max_growth_factor} is: {n_max} which gives {2**n_max}")
    for n in range(1, n_max):
        print(f"Day on which growth factor is {2**n}: {day(2**n)}")
        print(f"-----------Time to double: {day(2**n) - day(2**(n-1))}---------")
doubling_time()
```

```python
import numpy as np
import numpy.random as rnd

sprinkling = []
sprinklings = []
for j in range(10):
    for _ in range(3):
        sprinkling.append((rnd.random(), rnd.random(), rnd.random()))
    sprinklings.append(sprinkling)
    sprinkling = []
```

```python
sps = np.array(sprinklings)
sps_all_rs = sps[:,:,2]
print(sps_all_rs)

sps_mean_rs = np.mean(sps_all_rs, axis=1)
print(sps_mean_rs)
```

```python
means = [np.mean([r for _,_,r in sprinkling]) for sprinkling in sprinklings]
print(np.round(means, 2))
```
