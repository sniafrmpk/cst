## while loops

The syntax for a `while` is

```julia
while *condition*
    *loop body*
end
```

For example, we could use `while` to count or to iterate over an array.

````julia
n = 0
while n < 10
    n += 1
    println(n)
end
n
````

````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

i = 1
while i <= length(myfriends)
    friend = myfriends[i]
    println("Hi $friend, it's great to see you!")
    i += 1
end
````

## for loops

The syntax for a `for` loop is

```julia
for *var* in *loop iterable*
    *loop body*
end
```

We could use a for loop to generate the same results as either of the examples above:

````julia
for n in 1:10
    println(n)
end
````

````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

for friend in myfriends
    println("Hi $friend, it's great to see you!")
end
````

Now let's use `for` loops to create some addition tables, where the value of every entry is the sum of its row and column indices. <br>

Note that we iterate over this array via column-major loops in order to get the best performance. More information about fast indexing of multidimensional arrays inside nested loops can be found at https://docs.julialang.org/en/v1/manual/performance-tips/#Access-arrays-in-memory-order,-along-columns-1

First, we initialize an array with zeros.

````julia
m, n = 4, 5
A = fill(0, (m, n))
````

````julia
for j in 1:n
    for i in 1:m
        A[i, j] = i + j
    end
end
A
````

Here's some syntactic sugar for the same nested `for` loop

````julia
B = fill(0, (m, n))
````

````julia
for j in 1:n, i in 1:m
    B[i, j] = i + j
end
B
````

The more "Julia" way to create this addition table would have been with an *array comprehension*.

````julia
C = [i + j for i in 1:m, j in 1:n]
````

````julia
function recur_factorial(n)
    if n == 0
        return 1
    else
        return n * recur_factorial(n - 1)
    end
end
@assert recur_factorial(5) == factorial(5)

````

````julia
function recur_sum(n)
    if n == 0
        return 0
    end
    a = 0
    for i in 1:n
        a = a + i
    end
    return a
end
@assert recur_sum(5) == 15

b = 1e5
println("b = $b")
````

Writing my own insertion sort!

````julia
# The code should be able to sort a small array in an ascending order from left to right. The first j-1 elements should always be sorted
# and the jth element then finds its position within them by being moved to a lefter position until the element to its left is less than it.

function insertion_sort_v1(A)
    for j in 2:length(A)
        key = A[j]
        i = j - 1

        while i > 0
            if A[i] > key  # put this condition in the while loop because otherwise the while loop keeps on 
                           # running even after the correct spot for key has been found
                A[i + 1] = A[i]
                A[i] = key # this is not needed because if key needs to be moved two spots then on the second 
                           # run of the loop you will rewrite the key assignment of the previous run through 
                           # A[i+1] = A[i]. Instead put A[i + 1] = key after the while loop.
                
            end
            i -= 1
        end
    end
    return A
end

function insertion_sort_updated(A)
    for j in 2:length(A)
        key = A[j]
        i = j - 1
        while i > 0 && A[i] > key
            A[i + 1] = A[i]
            i -= 1
        end
        A[i + 1] = key
    end
    return A
end
````

````julia
using BenchmarkTools, Random

N = 50_000
data = rand(1:1_000_000, N)

@btime begin v = copy($data); insertion_sort_v1(v); end
@btime begin v = copy($data); insertion_sort_updated(v); end
````
